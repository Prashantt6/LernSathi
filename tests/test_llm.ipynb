```python
import ollama
import time

start = time.time()

response = ollama.chat(
    model="qwen3:4b",
    messages=[
        {
            "role": "system",
            "content": (
                "Du bist ein freundlicher deutscher Sprachlehrer. "
                "Antworte immer kurz, maximal zwei Sätze."
            ),
        },
        {
            "role": "user",
            "content": "Hallo! Wie geht es dir?"
        }
    ],
    options={
        "num_ctx": 2048,
        "temperature": 0.7,
    }
)

elapsed = time.time() - start

print(response["message"]["content"])
print(f"\nTime: {elapsed:.2f} seconds")
```

```python
import ollama
import time

start = time.time()

response = ollama.chat(
    model="qwen3:1.7b",
    messages=[
        {
            "role": "system",
            "content": (
                "Du bist ein freundlicher deutscher Sprachlehrer. "
                "Antworte kurz und natürlich auf Deutsch."
            ),
        },
        {
            "role": "user",
            "content": "Hallo! Wie geht es dir?"
        }
    ],
    options={
        "num_ctx": 2048,
        "num_predict": 80,
        "temperature": 0.7,
    }
)

print("FULL RESPONSE:")
print(response)

print("\nMESSAGE:")
print(response["message"])

print("\nCONTENT:")
print(repr(response["message"]["content"]))

print(f"\nTime: {time.time() - start:.2f}s")
```
