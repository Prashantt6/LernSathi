```python
from pathlib import Path
import subprocess
import sys

project_root = Path.cwd().parent

model_path = project_root / "models" / "tts" / "de_DE-thorsten-medium.onnx"
output_path = project_root / "audio" / "output" / "test.wav"

output_path.parent.mkdir(
    parents=True,
    exist_ok=True,
)

text = """Ah, hallo!  Dein Satz ist fast richtig, aber es fehlt ein Komma nach „Hallo“. Richtig wäre: *„Hallo, ich möchte Deutsch lernen“*.  
Wie möchten wir heute Deutsch lernen? Möchtest du heute einfach Wörter lernen oder einen kleinen Satz bauen? """

print("Project root:", project_root)
print("Model path:", model_path)
print("Model exists:", model_path.exists())

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "piper",
        "--model",
        str(model_path),
        "--output_file",
        str(output_path),
    ],
    input=text.encode("utf-8"),
    capture_output=True,
)

print("Return code:", result.returncode)

if result.stdout:
    print("STDOUT:")
    print(result.stdout.decode("utf-8", errors="replace"))

if result.stderr:
    print("STDERR:")
    print(result.stderr.decode("utf-8", errors="replace"))

print("Audio generated:", output_path.exists())
```
