```python
import whisper

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", device)

model= whisper.load_model("small", device=device)

print("Whisper is loaded")
```

```python
audio_path = "../audio/input/german_test.mp3"

result = model.transcribe(
    audio_path,
    language="de"
)

print(result["text"])
```

```python
import ollama

messages = [
    {
        "role": "system",
        "content": (
            "Du bist ein freundlicher deutscher Sprachlehrer. "
            "Führe eine natürliche Unterhaltung auf Deutsch."
        )
    }
]

messages.append({
    "role": "user",
    "content": "Ich heiße Prashant und studiere Computer Engineering."
})

response = ollama.chat(
    model="qwen3:4b",
    messages=messages
)

print("AI:", response["message"]["content"])

messages.append({
    "role": "assistant",
    "content": response["message"]["content"]
})
```
