```python
from ai.speech.stt import SpeechToText
from ai.speech.tts import TextToSpeech
from ai.llm.model import GermanChatbot

from IPython.display import Audio, display
```

```python
stt = SpeechToText("small")
tts = TextToSpeech()
chatbot = GermanChatbot()

print("All AI components loaded!")
```

```python
def run_pipeline(audio_path: str):

    # -------------------------
    # 1. Speech → Text
    # -------------------------
    print(" Transcribing...")

    user_text = stt.transcribe(audio_path)

    print("\n You:")
    print(user_text)

    # -------------------------
    # 2. Text → LLM
    # -------------------------
    print("\n Thinking...")

    messages = [
        {
            "role": "user",
            "content": user_text
        }
    ]

    ai_response = chatbot.generate_response(messages)

    print("\n Tutor:")
    print(ai_response)

    # -------------------------
    # 3. Text → Speech
    # -------------------------
    print("\n Generating speech...")

    audio_output = tts.synthesize(
        ai_response,
    )

    print("\n Pipeline complete!")

    return {
        "user_text": user_text,
        "ai_response": ai_response,
        "audio_path": audio_output,
    }
```

```python
result = run_pipeline(
    "../audio/input/german_test.mp3"
)
```

```python
conversation_history = []
```

```python
def run_pipeline(audio_path: str, conversation_history: list):

    # -------------------------
    # 1. Speech → Text
    # -------------------------
    print(" Transcribing...")

    user_text = stt.transcribe(audio_path)

    print("\n You:")
    print(user_text)

    # Add user message in history
    conversation_history.append({
        "role": "user",
        "content": user_text
    })

    # -------------------------
    # 2. Text → LLM
    # -------------------------
    print("\n Thinking...")

    ai_response = chatbot.generate_response(conversation_history)

    print("\n Tutor:")
    print(ai_response)

    # Add AI response
    conversation_history.append({
        "role": "assistant",
        "content": ai_response
    })
    # -------------------------
    # 3. Text → Speech
    # -------------------------
    print("\n Generating speech...")

    audio_output = tts.synthesize(
        ai_response,
    )

    print("\n Pipeline complete!")

    return {
        "user_text": user_text,
        "ai_response": ai_response,
        "audio_path": audio_output,
    }
```

```python
result = run_pipeline(
    "../audio/input/german_test.mp3",
    conversation_history
)
```
